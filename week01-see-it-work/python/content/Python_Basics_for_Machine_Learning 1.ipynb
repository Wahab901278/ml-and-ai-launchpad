## 2. Arithmetic Operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | addition | `5 + 2` -> `7` |
| `-` | subtraction | `5 - 2` -> `3` |
| `*` | multiplication | `5 * 2` -> `10` |
| `/` | division, always returns float | `5 / 2` -> `2.5` |
| `//` | floor division, drops the remainder | `5 // 2` -> `2` |
| `%` | modulus, gives the remainder | `5 % 2` -> `1` |
| `**` | exponent (power) | `5 ** 2` -> `25` |

```python
val1 = 15
val2 = 7

print(val1 + val2)   # addition
print(val1 - val2)   # subtraction
print(val1 * val2)   # multiplication
print(val1 / val2)   # division
print(val1 // val2)  # floor division
print(val1 % val2)   # modulus (remainder)
print(val1 ** val2)     # exponent
```

Python follows normal operator precedence (power first, then multiply/divide, then add/subtract). Use brackets to make the order explicit and easy to read.

```python
# without brackets, precedence rules decide the order (PEDMAS)
result = 4 + 5 * 2 ** 2 - 8 * 6
print(result)

# with brackets, the order is explicit
result = (4 + 5) * (2 ** 2) - (8 * 6)
print(result)
```

**Real world link:** arithmetic operators are used to normalize data (`(value - min) / (max - min)`), compute averages for reports, and calculate error metrics such as `(actual - predicted) ** 2` used in mean squared error.

## 3. Comparison (Relational) Operators

Comparison operators compare two values and always return a boolean (`True` or `False`).

| Operator | Meaning |
|---|---|
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |
| `==` | equal to |
| `!=` | not equal to |

```python
x = 15
y = 10

print(x > y)    # greater than
print(x < y)    # less than
print(x >= 15)  # greater than or equal to
print(y <= 10)  # less than or equal to
print(x == y)   # equal to
print(x != y)   # not equal to
```

```python
# comparison also works on strings, based on alphabetical order
print("apple" == "apple ")
print("apple" == "banana")
print("apple" < "banana")
```

## 4. Logical Operators

Logical operators combine multiple boolean conditions.

- `and` returns `True` only if both conditions are `True`
- `or` returns `True` if at least one condition is `True`
- `not` flips a boolean value

```python
age = 20
has_id = True

print(age >= 18 and has_id)   # both must be true
print(age >= 18 or has_id)    # at least one must be true
print(not has_id)             # flips True to False
```

```python
# combining comparisons and logical operators
score = 72
print(score >= 60 and score <= 100)   # normal passing range
print(not (score < 60))               # same result, written differently
```

**Real world link:** logical operators are used to filter data, for example keeping only rows where `age >= 18 and country == "PK"`, or to build alert rules like `if temperature > 100 or pressure < 20`.

## 5. Conditional Statements (if, elif, else)

Conditional statements let a program take different actions depending on whether a condition is `True` or `False`.

```python
number = 7

if number > 0:
    print("positive number")
elif number < 0:
    print("negative number")
else:
    print("number is zero")
```

```python
# a discount example using if / elif / else
price = 250

if price < 100:
    discount = price * 0.05
elif price < 500:
    discount = price * 0.10
else:
    discount = price * 0.20

print("Discount amount:", discount)
```

```python
# nested if: an if statement inside another if statement
num = 18

if num >= 0:
    if num == 0:
        print("zero")
    else:
        print("positive number")
else:
    print("negative number")
```

**Real world link:** conditional statements are how a trained model turns a probability into a decision. A spam filter might use `if probability > 0.5: label = "spam" else: label = "not spam"`. Data cleaning code uses conditions too, for example `if value is missing: fill_with_average()`.

## 6. Loops

Loops repeat a block of code. `for` loops iterate over a known sequence (a range, a list, a string). `while` loops repeat as long as a condition stays `True`.

```python
# for loop over a range of numbers
for number in range(1, 6):
    print(number)
```

```python
for number in range(1,101):
    print('5 x ',number,'=',5*number)
```

```python
# for loop over a list
ingredients = ["milk", "sugar", "tea leaves"]

for ingredient in ingredients:
    print(f"Add {ingredient}")
```

```python
# while loop, repeats until the condition becomes False
count = 0

while count < 5:
    print(count)
    count += 1 # count = count +1
```

`break` stops the loop completely. `continue` skips the rest of the current cycle and moves to the next one.

```python
numbers = [3, 7, 12, 5, 9]

for num in numbers:
    if num % 2 == 0:
        print("found an even number:", num)
        break
else:
    print("no even number found")
```

```python
# continue skips just this cycle, the loop keeps going
for num in range(1, 11):
    if num % 2 == 0:
        continue   # skip even numbers
    print(num)
```

**Real world link:** loops are used to go through every row of a dataset during cleaning, to run one training epoch at a time while training a model, and to process data in batches instead of all at once.

## 7. Functions

A function is a reusable block of code that takes input, does some work, and (usually) returns output. Functions avoid repeating the same code again and again.

```python
def add_numbers(a, b):
    return a + b

```

```python
result = add_numbers(5, 3)
print(result)
```

```python
# default arguments, used when the caller does not provide a value
def greet(name="friend"):
    return f"Hello, {name}!"

print(greet("Sara"))
print(greet())
```

```python
# default arguments, used when the caller does not provide a value
def greet_2(name):
    return f"Hello, {name}!"

print(greet_2("Sara"))
print(greet_2())
```

```python
def add_numbers(a,b,c,d,e,f,g):
    return a+b+c

add_numbers(5,3,7,5,6,8)
```

```python
def add_numbers(**kwargs):
    print(kwargs.values())

add_numbers(a=7,b=2,c=1)
```

```python
# *args collects any number of positional arguments into a tuple
def add_all(*args):
    total = 0
    for value in args:
        total += value # total = total +value
    return total

print(add_all(3, 4, 7, 8))
```

```python
# **kwargs collects any number of keyword arguments into a dictionary
def multiply_all(**kwargs):
    result = 1
    for value in kwargs.values():
        result *= value
    return result

print(multiply_all(a=2, b=3, c=4))
```

### Variable scope

A variable created inside a function is local, it only exists inside that function. A variable created outside every function is global, it can be read from anywhere in the file.

```python
total_count = 0   # global variable

def increment():
    # global step
    step = 1      # local variable, only exists inside this function
    return total_count + step
print(total_count)
print(increment())
print(step)    #"this would raise an error, step does not exist out here
```

```python
add_numbers(99,1)
```

**Real world link:** functions are the building blocks of an ML pipeline, a `clean_data()` function, a `train_model()` function, and an `evaluate()` function are each written once and reused for every experiment instead of copy-pasting the same code.

## 8. Lambda Functions

A lambda function is a small, unnamed function written in a single line. It is used when a short function is needed for a brief moment, often passed into another function.

```python
# a normal function
def square(x):
    return x ** 2

# the same thing as a lambda function
# square_lambda = lambda x: x ** 2

print(square(5))
# print(square_lambda(5))
```

```python
lambda_variable=lambda inp1: inp1**2
```

```python
lambda_variable(8)
```

```python
# lambda is commonly used with sorted(), to control how items are compared
students = [("Ali", 82), ("Sara", 91), ("Bilal", 76)]

students_sorted = sorted(students, key=lambda student: student[1])
print(students_sorted)
```

```python
# lambda used with map(), to apply a small function to every item
prices = [100, 250, 400]
prices_with_tax = list(map(lambda price: price * 1.15, prices))
print(prices_with_tax)
```

**Real world link:** lambda functions are common in data analysis, for example `dataframe["col"].apply(lambda x: x.strip())` to clean every value in a column, or as the sort key when ranking model results by score.

## 9. Data Structures

Python has four built in data structures used constantly in data work: lists, tuples, sets and dictionaries.

### Lists

A list is ordered, mutable (changeable), and allows duplicate values.

```python
fruits = ["apple", "banana", "cherry"]

print(fruits[0])       # first item
print(fruits[-1])      # last item
print(fruits[1:])      # slicing, from index 1 to the end

fruits[1] = "blueberry"   # lists can be changed after creation
print(fruits)
```

### Tuples

A tuple is ordered like a list, but immutable, it cannot be changed after creation. Tuples are used for fixed, unchanging groups of values.

```python
point = (10, 20)
print(point[0], point[1])

# unpacking a tuple into separate variables
x, y = point
print(x, y)

point[0] = 99   # this would raise an error, tuples cannot be changed
```

```python
point.count(90)
```

```python
point.index(20)
```

```python
l1=list(point)
l1[1]=99
point=tuple(l1)
point
```

**Real world link:** tuples are used for things that should not change, like an `(image, label)` pair in a training dataset, or a function that returns more than one value, such as `(accuracy, loss)`.

### Sets

A set is unordered and only stores unique values, duplicates are automatically removed.

```python
tags = {"python", "ml", "ai", "python"}   # duplicate "python" is dropped
print(tags)

tags.add("data")
# print(tags)
tags.remove("ai")
print(tags)

set_a = {1, 2, 3}
set_b = {2, 3, 4}
print(set_a.union(set_b))          # all unique values from both sets
print(set_a.intersection(set_b))   # values present in both sets
print(set_a.difference(set_b))     # values only in set_a
```

```python
tags[1]
```

**Real world link:** sets are used to remove duplicate records from a dataset, or to build a vocabulary of unique words when processing text for natural language processing.

### Dictionaries

A dictionary stores data as key-value pairs. Keys must be unique, and each key maps to one value.

```python
student = {"name": "Ahmed", "age": 23, "course": "AI"}

print(student["name"])          # access by key
print(student.get("age"))# safer access, does not error if key is missing
# print(student.get("grade"))
# print(student["grade"])
print(student.get("grade", "N/A"))  # default value if key is missing

student["age"] = 24             # update a value
student["grade"] = "A"          # add a new key
print(student)
```

```python
for i in student:
    print('key:',i,'value:',student[i])
```

```python
for i in student.values():
    print(i)
```

```python
for i in student.items():
    print(i)
```

```python
# looping over a dictionary
for key, value in student.items():
    print(key, ":", value)

print(list(student.keys()))
print(list(student.values()))
```

```python
# dictionary comprehension
squares_map = {n: n ** 2 for n in range(1, 6)}
print(squares_map)
```

**Real world link:** dictionaries are used for configuration settings, for mapping categories to numbers during label encoding (for example `{"low": 0, "medium": 1, "high": 2}`), and for counting word frequency in text data.

## 10. File Handling

File handling lets a program read data from a file and write data back to one. The `with open(...) as file:` pattern is preferred because it closes the file automatically once done.

Common modes: `"r"` read, `"w"` write (overwrites the file), `"a"` append (adds to the end).

```python
# writing to a text file
# with open("example.txt", "w") as f1:
#     f1.write("Hello, world!\n")
#     f1.write("This is a Python file handling example.")

# reading the whole file back
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
with open("example.txt", "w") as f1:
    f1.write("Hello, I am abdul")
```

```python
for i in range(1,101):
    with open("example.txt", "a") as f1:
        f1.write("Hello, I am abdul 2\n")
```

```python
# reading a file line by line
with open("example.txt", "r") as file:
    for line in file:
        print(line.strip())
```

```python
# appending, adds new content without erasing what is already there
with open("example.txt", "a") as file:
    file.write("\nAppended line.")

with open("example.txt", "r") as file:
    print(file.read())
```

```python
import csv

# writing a small CSV file, a common dataset format
rows = [
    ["name", "age", "score"],
    ["Ahmed", 23, 88],
    ["Sara", 25, 91],
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(rows)

# reading the CSV file back
with open("students.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

**Real world link:** file handling is how a dataset gets loaded into a program in the first place (`.csv`, `.json`, `.txt`), and how trained model results, logs, or configuration get saved back to disk.
