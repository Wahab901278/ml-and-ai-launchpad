```python
def report(df,name="dataset"):
    n_rows,n_columns=df.shape
    n_numeric_columns=df.select_dtypes(include='number').shape[1]
    n_text_columns=df.select_dtypes(include='object').shape[1]

    print(f"{name}: {n_rows}rows, {n_columns} columns")
    print(f"numeric : {n_numeric_columns}")
    print(f"text columns : {n_text_columns}")
    print(f"memory: {df.memory_usage(deep=True).sum()/1e6} MB")
```

```python
report(raw,name='airbnb')
```

```python
def check_missing_value(df):
    df=pd.DataFrame(
            {
                "missing": df.isna().sum(),
                "missing_%": (df.isna().mean()*100).round(2)
            }
    )
    return df
```

```python


def create_bar_chart(
    x: str,
    y: str,
    title: str = "",
    xlabel: str = "",
    ylabel: str = "",
    orient: str = "v",
    palette: str = "viridis",
    figsize: tuple = (10, 6),
    save_path: str = None,
) -> plt.Axes:
    
    plt.bar(x,y)
    plt.xticks(rotation=70)
```
