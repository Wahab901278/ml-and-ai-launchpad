```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences=[
    'The food is good',
    'The food is bad',
    'The food is not good'
]
```

```python
voc_size=100000
```

```python
sentences
```

```python
one_hot_repr=[]
for words in sentences:
    one_hot_repr.append(one_hot(words,voc_size))
```

```python
one_hot_repr
```

```python
sentences
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
sent_length=5
padded_sent=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
```

```python
padded_sent
```

```python
help(pad_sequences)
```

```python
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,input_length=sent_length,output_dim=dim))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(padded_sent)
```
